````python
import wisemind_client
from wisemind_client import WiseMind, AsyncWiseMind, Model, ProfilePatient, WiseMindError
from importlib.metadata import version
print("wisemind-client", version("wisemind-client")); print(wisemind_client.__all__)
````

## 2. Server reachable
`/healthz` and `/api/config` need no token. `protocol: 2` is the JSON frame protocol this SDK speaks.

````python
import urllib.request, ssl, certifi
ctx = ssl.create_default_context(cafile=certifi.where())      # same CA bundle the SDK uses for wss://
http_base = BASE_URL.replace("ws://", "http://").replace("wss://", "https://").rstrip("/")
print(json.load(urllib.request.urlopen(f"{http_base}/healthz", timeout=10, context=ctx)))
cfg = json.load(urllib.request.urlopen(f"{http_base}/api/config", timeout=10, context=ctx)); print(cfg)
assert cfg["protocol"] == 2
````

## 3. Models and client
RA (Reasonable Mind) makes the decisions; EA (Emotional Mind) writes the dialogue. Any OpenAI-compatible endpoint, each with its own key and temperature. Here both are DeepSeek with different temperatures; mix providers freely.

````python
ra = Model(base_url=DS, model="deepseek-chat", api_key=API_KEY, temperature=0)
ea = Model(base_url=DS, model="deepseek-chat", api_key=API_KEY, temperature=0.8)
client = WiseMind(BASE_URL, access_token=ACCESS_TOKEN)
print("websocket:", client.url)
print("what the server receives for ra (key redacted):", {**ra.spec(), "api_key": "***"})
````

## 4. Refusals
All checked before any model call. The SDK raises `WiseMindError` with the server's `code`.

````python
import certifi
os.environ.setdefault("SSL_CERT_FILE", certifi.where())
def expect(label, fn):
    try: fn(); print(f"{label}: no error (unexpected)")
    except WiseMindError as e: print(f"{label}: [{e.code}] {e}")
deny = lambda q: "No, nothing like that."
expect("wrong token      ", lambda: WiseMind(BASE_URL, "wrong").run("mdd", ra, ea, deny))
expect("unknown tree     ", lambda: client.run("xyz", ra, ea, deny))
expect("private host     ", lambda: client.run("mdd", Model("http://127.0.0.1:9/v1", "x", "k"), ea, deny))
expect("temperature > 2  ", lambda: client.run("mdd", Model(DS, "deepseek-chat", API_KEY, temperature=5), ea, deny))
````

## 5. Scripted interview, turn by turn, verbose
The core loop. `verbose=True` prints each node decision as the server makes it. The replies come from the bundled `ProfilePatient` (answers from a VSP CSV with a model of your choice); replace it with anything that maps a question to a reply.

````python
patient = ProfilePatient(str(DATA / "example_vsp_mdd" / "MDD_0.csv"), model=ra)
t0 = time.time()
with client.start(tree="mdd", ra=ra, ea=ea, verbose=True) as session:
    for i, question in enumerate(session):
        answer = await patient(question)          # ProfilePatient is async; in a script use patient.sync(question)
        print(f"[{i:02d}] DOCTOR : {question[:110]}\n     PATIENT: {answer[:110]}")
        session.reply(answer)
r = session.result
print(f"\nDIAGNOSIS: {r.diagnosis} ({r.description})\nPATH: {' -> '.join(r.path)}\nturns={r.turns} backtracks={r.backtracks} wall={time.time()-t0:.0f}s")
assert r.diagnosis == "MDD"
````

## 6. The result object

````python
print("nodes     :", r.nodes)
print("transcript:", len(r.transcript), "exchanges; first:", r.transcript[0])
print("stats     :", r.stats)
print("timing    :", r.timing)
print("events    :", len(r.events), "frames;", {e["type"] for e in r.events})
r.save("MDD_0_result.json"); print("saved MDD_0_result.json")
````

## 7. One-call form with a hand-written patient

## 9. Mixed providers and the light role
RA and EA can be different providers. `light` (summariser, sanity checker) defaults to `ra`. Set `OPENAI_API_KEY` to run this cell with GPT-4o as the dialogue model; otherwise it is skipped.

````python
if os.getenv("OPENAI_API_KEY"):
    ea_gpt = Model("https://api.openai.com/v1", "gpt-4o", os.environ["OPENAI_API_KEY"], temperature=0.8)
    r3 = client.run("mdd", ra, ea_gpt, ProfilePatient(str(DATA / "example_vsp_mdd" / "NORMALBLUE_0.csv"), model=ra).sync, light=ra)
    print(r3.diagnosis, r3.stats["models_served"])
else:
    print("skipped: OPENAI_API_KEY not set")
````

## 10. Async client, interviews in parallel

````python
async def two():
    ac = AsyncWiseMind(BASE_URL, ACCESS_TOKEN)
    async def one(p):
        res = await ac.run("anx", ra, ea, ProfilePatient(str(p), model=ra))
        return p.stem, res.diagnosis, res.turns
    return await asyncio.gather(*(one(p) for p in [DATA / "example_vsp_anx" / "SOCIALANX_0.csv",
                                                     DATA / "example_vsp_anx" / "SPEPHOBIA_0.csv"]))
for stem, pred, turns in await two():
    gt = stem.rsplit("_", 1)[0]; print(f"{stem:16s} gt={gt:12s} pred={pred:12s} {'ok' if gt == pred else 'MISS'} turns={turns}")
````

## 12. Failure mid-interview
A dead provider key: the server ends the session and the SDK raises `WiseMindError(code='model_failure')` instead of hanging. The key is masked in the message.

````python
try:
    client.run("mdd", Model(DS, "deepseek-chat", "sk-invalid-key-000000000000"), ea, deny)
except WiseMindError as e:
    print(f"[{e.code}] {str(e)[:220]}")
````

## 13. Small accuracy batch (optional, ~10 min)
The embedded bipolar profiles, 3 in parallel, with the reference patient.

````python
files = sorted((DATA / "example_vsp_bp").glob("*.csv"))
async def batch():
    ac = AsyncWiseMind(BASE_URL, ACCESS_TOKEN); sem = asyncio.Semaphore(3)
    async def one(f):
        async with sem:
            gt = f.stem.rsplit("_", 1)[0]
            try:
                res = await ac.run("bp", ra, ea, ProfilePatient(str(f), model=ra)); return gt, res.diagnosis, res.turns
            except WiseMindError as e: return gt, f"ERROR {e.code}", 0
    return await asyncio.gather(*(one(f) for f in files))
rows = await batch()
for gt, pred, turns in rows: print(f"{gt:14s} -> {pred:14s} {'ok' if gt == pred else 'MISS'}  turns={turns}")
print(f"top-1: {sum(gt == pred for gt, pred, _ in rows)}/{len(rows)}")
````
